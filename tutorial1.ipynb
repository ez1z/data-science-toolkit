Create a numpy array of size 2 x 2

```python
import numpy as np
array = np.array([[1, 1], [2, 2]])
print(array)
```

Create a random numpy array of size 3 x 4

```python
random_arr = np.random.rand(3, 4)
print(random_arr)
```

Create a numpy array of ones and a numpy array of zeroes

```python
ones = np.ones((4, 4))
zeros = np.zeros((3,3))

print(ones)
print(zeros)
```

Create a random numpy array of size 6x3 and transpose it

```python
random_arr = np.random.rand(6, 3)
print(random_arr)
transposed_arr = random_arr.transpose()
print(transposed_arr)
```

Create two random numpy array of size n x p and p x m and then compute the dot product

```python
random_arr1 = np.random.rand(3, 4)
random_arr2 = np.random.rand(4, 3)
print(random_arr1)
print(random_arr2)
dot_product = np.dot(random_arr1, random_arr2)
print(dot_product)
```
